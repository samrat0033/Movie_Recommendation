### Importing all the necessary libraries

```python
#Libaries
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')
```

```python
#importing the movie datasets
movie = pd.read_csv("tmdb_5000_movies.csv")
movie.head(5)
```

### Now we have to merge this data set on the basis of "Title"

```python
movies= movie.merge(credit,on='title')
movies.head(1)
```

### Now we have to select those column thats are necessary for this model
required columns = gener, id, keywords,overview, title, cast, crew

```python
movies = movies[["movie_id","genres","keywords","overview","title","cast","crew"]]
```

```python
movies.head(1)
```

### Now we have to check where there is any null data set present in the data frame if present then we have to adjust that data set or remove it

```python
movies.isnull().sum()
```

```python
movies.dropna(inplace = True)
```

```python
movies.iloc[0]['genres']
```

Now we have to reomove all the unnecessary data from this genres column


```python
# [{"id": 28, "name": "Action"}, {"id": 12, "name": "Adventure"}, {"id": 14, "name": "Fantasy"}, {"id": 878, "name": "Science Fiction"}]
# ie. ["Action","Adventature","Fantasy","sience fiction"]
```

```python
import ast
def convert(obj):
  l=[]
  #obj is a string so we have to convert this into a list
  #for that we usw ast.literal_eval function
  for i in ast.literal_eval(obj):
        l.append(i['name'])
  return l
```

```python
convert('[{"id": 28, "name": "Action"}, {"id": 12, "name": "Adventure"}, {"id": 14, "name": "Fantasy"}, {"id": 878, "name": "Science Fiction"}]')
```

```python
movies.genres = movies.genres.apply(convert)
```

```python
movies.head(2)
```

```python
#we will do the same for keyword column
movies.keywords = movies.keywords.apply(convert)
```

```python
movies.head(1)
```

we have to do the same for the cast but we will take only 1st three cast name

```python
movies.cast[0]
```

```python
#ie. [Sam Worthington, Zoe Saldana, Sigourney Weaver]
```

```python
import ast
def select_cast(obj):
  counter=0
  l=[]
  #obj is a string so we have to convert this into a list
  #for that we usw ast.literal_eval function
  for i in ast.literal_eval(obj):
    if counter!=3:
      l.append(i["name"])
      counter+=1
    else:
      break
  return l
```

```python
movies.cast = movies.cast.apply(select_cast)
```

```python
movies.head(1)
```

we will do the same for crew but we will take just director name

```python
movies.crew[0]
```

```python
#[James Cameron]
```

```python
def fetch_director(obj):
  l=[]
  for i in ast.literal_eval(obj):
    if i['job']=='Director':
      l.append(i['name'])
      break
  return l
```

```python
movies.crew = movies.crew.apply(fetch_director)
```

```python
movies.head(1)
```

Now we have to remove spaces from words.
ie. Science Fiction to ScienceFiction

```python
movies.genres = movies.genres.apply(lambda x: [i.replace(" ","") for i in x])
movies.cast = movies.cast.apply(lambda x: [i.replace(" ","") for i in x])
movies.keywords = movies.keywords.apply(lambda x: [i.replace(" ","") for i in x])
movies.crew = movies.crew.apply(lambda x: [i.replace(" ","") for i in x])
```

```python
movies.head(1)
```

```python
movies.overview = movies.overview.apply(lambda x: x.split() if isinstance(x, str) else []) # this will convert overview into list
```

Now we will merge this four colum and make them tag

```python
movies["tags"] = movies["overview"]+movies["genres"]+movies["keywords"]+movies["cast"]+movies["crew"]
```

```python
movies.head(1)
```

```python
df = movies[["movie_id","title","tags"]]
```

```python
df.head(2)
```

Now we will convert this list(tags) into string

```python
df["tags"] = df["tags"].apply(lambda x:" ".join(x))
```

```python
df.head(5)
```

```python
df.tags[0]
```

We willl write a function that will convert the words into their preform i.e. loving-> love, loved-> love, lovely-> love. this will hepl our model to understand the tags more accurately.


```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
  y=[]
  for i in text.split():
    y.append(ps.stem(i))
  return " ".join(y)
```

```python
df["tags"]=df["tags"].apply(stem)
```

```python
df.tags = df.tags.apply(lambda x: x.lower())
```

```python
df.tags[0]
```

Now we have to vectorize our data because it is all text. For that we will use Sklearn count Vectorizer moduel

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=5000,stop_words="english")
```

```python
vector = cv.fit_transform(df["tags"]).toarray()
```

```python
vector
```

```python
vector.shape
```

```python
vector[0]
```

```python
cv.get_feature_names_out()
```

Now what we will do is we calculate the cosine distance of all the data point to find the similarity between the data sets.

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity = cosine_similarity(vector)
```
